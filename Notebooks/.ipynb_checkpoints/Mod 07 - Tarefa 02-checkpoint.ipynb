````python
X = df.drop(columns = 'mau_True', axis=1)
y = df.mau_True
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
````

#### 2) Vamos para o modelo:

1. Defina um objeto com a função da árvore de decisão vista em aula.

````python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
````

````python
model.fit(X_train, y_train)
````

````python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
````

````python
plt.figure(figsize=(40,20))
plot_tree(model, feature_names=X_train.columns, class_names=["reprovados", "aprovados"], filled=True, max_depth=3)
plt.show()
````

````python
plt.figure(figsize=(40, 20))  # valores maiores que (20,10)
plot_tree(model, feature_names=X_train.columns, class_names=["reprovados", "aprovados"], filled=True)
plt.show()
````
